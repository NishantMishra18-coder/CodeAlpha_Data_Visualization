```python
import pandas as pd

df = pd.read_csv(
    "Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python

# Step 3: Understand the Dataset

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape (Rows, Columns):")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python

# Step 4: Data Quality Check

print("Missing Values in Each Column:")
display(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
display(df.dtypes)
```

```python

# Step 5: Basic Sales Analysis

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())

print("\nSales by Category:")
display(
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

# Visualize Sales by Category
df.groupby("Category")["Sales"].sum().sort_values().plot(
    kind="bar",
    title="Total Sales by Category",
    xlabel="Category",
    ylabel="Sales",
    figsize=(8, 5)
)

plt.tight_layout()
plt.show()
```

```python

# Step 6: Sales and Profit Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Sales and Profit by Category
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

print("Sales and Profit by Category:")
display(category_analysis)

# 2. Sales and Profit by Region
if "Region" in df.columns:
    region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()

    print("Sales and Profit by Region:")
    display(region_analysis)

# 3. Sales and Profit Visualization
category_analysis.plot(kind="bar", figsize=(9, 5))

plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Step 7: Monthly Sales Trend

date_col = next(
    (col for col in df.columns
     if col.lower() in ["order date", "order_date", "date"]),
    None
)

if date_col:
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

    monthly_sales = (
        df.dropna(subset=[date_col])
          .set_index(date_col)["Sales"]
          .resample("ME")
          .sum()
    )

    monthly_sales.plot(figsize=(12, 5), marker="o")

    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.tight_layout()
    plt.show()
else:
    print("Date column nahi mila. Available columns:", df.columns.tolist())
```

```python

# Step 8: Top 10 Products by Sales

product_col = next(
    (col for col in df.columns
     if col.lower() in ["product name", "product_name", "product"]),
    None
)

if product_col:
    top_products = (
        df.groupby(product_col)["Sales"]
          .sum()
          .sort_values(ascending=False)
          .head(10)
    )

    display(top_products.reset_index())

    top_products.sort_values().plot(
        kind="barh", figsize=(10, 6)
    )

    plt.title("Top 10 Products by Sales")
    plt.xlabel("Total Sales")
    plt.tight_layout()
    plt.show()
else:
    print("Product column nahi mila.")
```

```python

# Step 9: Correlation Analysis

import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(10, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python

# Step 10: Outlier Detection using IQR

for col in ["Sales", "Profit"]:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[
            (df[col] < lower) | (df[col] > upper)
        ]

        print(f"\n{col}")
        print("Lower limit:", round(lower, 2))
        print("Upper limit:", round(upper, 2))
        print("Outlier count:", len(outliers))
```

```python

# Step 11: Business Summary

print("========== BUSINESS SUMMARY ==========")

print("\n1. Overall Sales:", round(df["Sales"].sum(), 2))
print("2. Overall Profit:", round(df["Profit"].sum(), 2))

if "Category" in df.columns:
    category_sales = df.groupby("Category")["Sales"].sum()
    print("\n3. Highest Sales Category:", category_sales.idxmax())
    print("4. Lowest Sales Category:", category_sales.idxmin())

if "Region" in df.columns:
    region_profit = df.groupby("Region")["Profit"].sum()
    print("\n5. Most Profitable Region:", region_profit.idxmax())
    print("6. Lowest Profit Region:", region_profit.idxmin())

print("\n======================================")
```

```python

import pandas as pd

filename = csv_files[0]

try:
    df = pd.read_csv(filename, encoding='cp1252')
except UnicodeDecodeError:
    df = pd.read_csv(filename, encoding='latin1')

print("✅ Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total sales by category
category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

# Create the chart
plt.figure(figsize=(9, 5))
ax = sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title('Total Sales by Product Category', fontsize=15)
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=0)

# Add sales values on top of bars
for bar in ax.patches:
    ax.annotate(
        f'${bar.get_height():,.0f}',
        (bar.get_x() + bar.get_width() / 2,
         bar.get_height()),
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.tight_layout()
plt.savefig('chart1_sales_by_category.png', dpi=300,
            bbox_inches='tight')
plt.show()

print("Sales by Category:")
display(category_sales.to_frame(name='Total Sales ($)'))
```

```python

# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    errors='coerce'
)

# Remove invalid dates for this chart
date_df = df.dropna(subset=['Order Date']).copy()

# Calculate monthly sales
date_df['Year-Month'] = (
    date_df['Order Date'].dt.to_period('M')
)

monthly_sales = (
    date_df.groupby('Year-Month')['Sales']
    .sum()
    .sort_index()
)

# Create line chart
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o',
    linewidth=2
)

plt.title('Monthly Sales Trend', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    'chart2_monthly_sales_trend.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python

# Calculate total sales by region
region_sales = (
    df.groupby('Region')['Sales']
    .sum()
    .sort_values(ascending=False)
)

# Create bar chart
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title('Total Sales by Region', fontsize=15)
plt.xlabel('Region')
plt.ylabel('Total Sales ($)')

# Display values on bars
for bar in ax.patches:
    ax.annotate(
        f'${bar.get_height():,.0f}',
        (bar.get_x() + bar.get_width() / 2,
         bar.get_height()),
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.tight_layout()
plt.savefig(
    'chart3_sales_by_region.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()

print("Regional Sales Summary:")
display(region_sales.to_frame(name='Total Sales ($)'))
```

```python

# Calculate sales by product
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .nlargest(10)
    .sort_values(ascending=True)
)

# Create horizontal bar chart
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title('Top 10 Products by Total Sales', fontsize=16)
plt.xlabel('Total Sales ($)')
plt.ylabel('Product Name')

# Add values to bars
for bar in ax.patches:
    ax.annotate(
        f'${bar.get_width():,.0f}',
        (bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(5, 0),
        textcoords='offset points',
        va='center',
        fontsize=9
    )

plt.tight_layout()
plt.savefig(
    'chart4_top10_products.png',
    dpi=300,
    bbox_inches='tight'
)
plt.show()

print("Top 10 Products:")
display(top_products.sort_values(ascending=False).to_frame(
    name='Total Sales ($)'
))
```
